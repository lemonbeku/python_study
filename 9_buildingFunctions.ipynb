## Difference Between Arguments and Parameters
**Parameter** is the **variable defined** in a function's declaration, while an **argument** is the **actual value** passed to that function when it is called.
Example 1:

````python
# Parameters (Placeholders)
def add_numbers(x, y):
  return x + y

# Arguments (Actual Value)
result = add_numbers(5, 3)
````

Example 2:

````python
def greet(lang):
  if lang == 'es':
    print("Hola")
  elif lang == 'fr':
    print("Bonjour")
  else:
    print('Hello')

greet('en')
````

## Return Values
Often a function will take its arguments, do some computation, and return a value to be used as the value of the function call in the calling expression. It has two purposes:
1. Stops the function execution.
2. Sends a value back to where the function was called.
Example:

````python
def greet():
  return "Hello"
print(greet(), "Glenn")
print(greet(), "Sally")
````

Execution
1. greet() is called
2. Function executes return "Hello"
3. Function stops immediately
4. "Hello" is sent back to the print() statement

- A fruitful function is one that produces a result (or return value)
- The return statement ends the function execution and sends back the result of the function.

````python
def greet(lang):
  if lang == 'es':
    return 'Hola'
  elif lang == 'fr':
    return 'Bonjour' # we're done at this point if its fr
  else:
    return 'Hello'
print(greet('en'), 'Glenn')
print(greet('es'), 'Sally')
print(greet('fr'), 'Michael')
````

## Return Important Rule
When Python encounters a `return` statement:

````python
def example():
  print("Start")
  return("Done")
  print("This never runs")
example()
````

The statement after `return` is never executed.

## Understanding `max()`
Consider:

````python
big = max("Hello World")
````

Step-by-step
1. Python calls max().
2. "Hello World" is passed into the function.
3. max() examines all characters
```python
H e l l o   W o r l d
```
4. It finds the largest character alphabetically:
```python
w
```
5. max() executes:
```python
return "W"
```
6. The returned value replaces the function call. So, the statement becomes:
```python
big = "W"
```

## Functions with Multiple Parameters
A function can accept more than one parameter.
Example:

````python
# Create
def addtwo(a, b):
  added = a + b
  return added

# Call
x = addtwo(3, 5)
print(x)
````

Parameter Rules
- Number of arguments must match number of parameters.
- Arguments are matched by position (order matters).
Example

````python
addtwo(3, 5)
````

| --------- | -------- |
| a         | 3        |
| b         | 5        |
